**Unit 1 Q1: Probability density |Y(l,m)|^2 of spherical harmonics**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import lpmv
from math import factorial


def density(l, m, theta):
    norm = (2 * l + 1) / (4 * np.pi) * factorial(l - m) / factorial(l + m)
    return norm * lpmv(m, l, np.cos(theta)) ** 2


theta = np.linspace(0, 2 * np.pi, 400)

plt.figure(figsize=(6, 6))
ax = plt.subplot(projection="polar")
ax.set_theta_zero_location("N")
for l, m in [(1, 0), (1, 1)]:
    ax.plot(theta, density(l, m, theta), label=f"l={l}, m={m}")
ax.set_title("Probability density for l = 1")
ax.legend(loc="lower right")

plt.figure(figsize=(6, 6))
ax = plt.subplot(projection="polar")
ax.set_theta_zero_location("N")
for l, m in [(2, 0), (2, 1), (2, 2)]:
    ax.plot(theta, density(l, m, theta), label=f"l={l}, m={m}")
ax.set_title("Probability density for l = 2")
ax.legend(loc="lower right")

plt.show()
```
