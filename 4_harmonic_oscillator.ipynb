**Unit 2 Q1: Harmonic potential (units: hbar = m = omega = 1)**

```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000
x = np.linspace(-6, 6, N)
dx = x[1] - x[0]
V = 0.5 * x ** 2

H = np.diag(1 / dx ** 2 + V)
H = H + np.diag(-0.5 / dx ** 2 * np.ones(N - 1), 1)
H = H + np.diag(-0.5 / dx ** 2 * np.ones(N - 1), -1)

E, psi = np.linalg.eigh(H)

print("n    Numerical    Analytical (n + 1/2)")
for n in range(4):
    print(n, "   ", round(E[n], 4), "     ", n + 0.5)

ground = (1 / np.pi) ** 0.25 * np.exp(-x ** 2 / 2)
exact = [ground, np.sqrt(2) * x * ground, (2 * x ** 2 - 1) / np.sqrt(2) * ground]

plt.figure(figsize=(8, 5))
for n in range(3):
    numerical = psi[:, n] / np.sqrt(np.sum(psi[:, n] ** 2) * dx)
    if np.sum(numerical * exact[n]) < 0:
        numerical = -numerical
    plt.plot(x, numerical, label=f"n={n} numerical")
    plt.plot(x[::40], exact[n][::40], "o", label=f"n={n} analytical")

plt.xlabel("x")
plt.ylabel("Wavefunction")
plt.title("Harmonic oscillator")
plt.legend()
plt.grid(True)
plt.show()
```
