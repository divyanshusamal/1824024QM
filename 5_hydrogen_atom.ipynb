**Unit 3 Q1: s-wave Schrodinger equation for the hydrogen atom**

```python
import numpy as np
import matplotlib.pyplot as plt

e = 3.795
hc = 1973.0
mc2 = 0.511e6
c = hc ** 2 / (2 * mc2)

N = 1000
r_max = 20.0
dr = r_max / N
r = np.arange(1, N + 1) * dr
V = -e ** 2 / r

H = np.diag(2 * c / dr ** 2 + V)
H = H + np.diag(-c / dr ** 2 * np.ones(N - 1), 1)
H = H + np.diag(-c / dr ** 2 * np.ones(N - 1), -1)

E, u = np.linalg.eigh(H)

print("State      Numerical (eV)    Exact (eV)")
print("Ground     ", round(E[0], 3), "          ", -13.6)
print("1st excited", round(E[1], 3), "          ", round(-13.6 / 4, 3))

plt.figure(figsize=(8, 5))
for n in range(2):
    wave = u[:, n] / np.sqrt(np.sum(u[:, n] ** 2) * dr)
    if wave[5] < 0:
        wave = -wave
    plt.plot(r, wave, label=f"n={n + 1}, E={E[n]:.3f} eV")

plt.xlabel("r (Angstrom)")
plt.ylabel("u(r)")
plt.title("Hydrogen atom radial wavefunctions")
plt.legend()
plt.grid(True)
plt.show()
```
